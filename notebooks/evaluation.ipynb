## Import Libraries

```python
import sys
import pandas as pd
sys.path.append('..')
from src.faiss_retriever import ComplaintRetriever, RAGPromptBuilder, RAGGenerator
```

## Initialize RAG Components

```python
retriever = ComplaintRetriever(faiss_index_path='../vector_store/faiss.index',
                               metadata_csv_path='../vector_store/chunks_metadata.csv')
prompt_builder = RAGPromptBuilder()
generator = RAGGenerator()
```

## Define Representative Evaluation Questions

```python
evaluation_questions = [
    "Why are customers dissatisfied with credit cards?",
    "What issues do customers report about money transfers?",
    "Are there frequent complaints about savings accounts?",
    "What problems do users face with personal loans?",
    "Do customers report fraud-related issues in credit cards?",
    "What are common billing-related complaints?",
    "Are delays a major issue in financial services?",
]
```

## Run the RAG Pipeline for Each Question

```python
evaluation_results = []
for question in evaluation_questions:
    # Step 1: Retrieve relevant chunks
    retrieved_chunks = retriever.retrieve(question=question)
    # Step 2: Build prompt
    prompt = prompt_builder.build_prompt(retrieved_chunks=retrieved_chunks, question=question)
    # Step 3: Generate answer
    generated_answer = generator.generate_answer(prompt=prompt)
    # Store results (show only 1–2 sources)
    evaluation_results.append({
        'Question': question,
        "Generated Answer": generated_answer,
        "Retrieved Sources":[chunk['chunk_text'][:200]+"..." for chunk in retrieved_chunks[:2]],
        "Quality Score (1-5)": "",
        "Comments / Analysis": ""
    })
```

## Create Evaluation Table (Markdown-Ready)

```python
evaluation_df = pd.DataFrame(evaluation_results)
evaluation_df
```
